```python
daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,
2500, 2200, 3100, 2400, 3900, 5600, 4700] # 14 days, KES

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2



# qty = kg in stock, reorder = reorder level (kg), cost/price = KES per kg

stock = {
"Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
"Onions":  {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
"Sukuma":  {"qty": 9,"reorder": 15, "cost": 30, "price": 50},
"Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
"Avocado": {"qty": 12, "reorder": 10, "cost": 80, "price": 140},

}

total = sum (daily_sales)
print(f'Total : KES {total:,.0f}')

avg = sum(daily_sales) / len(daily_sales)
print(f'Average : KES {avg:,.0f}')

#high and low velues in daily sales

def high_low(daily_sales):
  return min(daily_sales), max(daily_sales)

highest_sales = max(daily_sales)
lowest_sales = min(daily_sales)
print(f'Highest sales : KES {highest_sales:,.0f}')
print(f'Lowest sales : KES {lowest_sales:,.0f}')

best_day = daily_sales.index(highest_sales)
best_week = (best_day// 7 ) +1
print(f'Best day : {days[best_day]} , position:{best_day}, week{best_week}')


worst_day= daily_sales.index(lowest_sales)
worst_week  = (worst_day // 7) +1
print(f'Worst day: {days[worst_day]} , position :{worst_day}, week{worst_week}')

#WEEK TOTALS

def week_totals(sales):


  week1 = sum(daily_sales[0:7])
  week2 = sum(daily_sales[7:14])
  return week1, week2
week1 ,week2 = week_totals(daily_sales)

print(f'Week 1 total : KES {week1:.0f}')
print(f'week 2 total :KES {week2:.0f}')

growth_percentage = ((week2 - week1) / week1) * 100
print(f'Growth percentage : {growth_percentage:.2f}%')

def clasify_day(amount,average):#clasify daily sales as strong,weak,or normal
  if amount > average* 1.20:
    return "strong"
  elif amount < average * 0.80:
    return "weak"
  else:
    return "Normal"
#dictionary to count each category
category = {
    'strong':0,
    'weak':0,
    'Normal':0
}

#for everyday sales

for amount in daily_sales:
  result = clasify_day(amount, avg)
  category[result] += 1
print( 'daily sale category:')
for category, count in category.items():
  print(f'{category}: {count} days')

weekday_sales =[]
weekend_sales = []

for i in range(len(daily_sales)) :
  if days[i] in ['Mon', 'Tue', 'Wed', 'Thu', 'Fri']:
    weekday_sales.append(daily_sales[i])
  else:
    weekend_sales.append(daily_sales[i])

weekday_average = sum(weekday_sales) / len(weekday_sales)
weekend_average = sum(weekend_sales) / len(weekend_sales)

difference = ((weekend_average-weekday_average)/weekday_average)*100

print(f'Weekday average : KES {weekday_average:.0f}')
print(f'Weekend average : KES {weekend_average:.0f}')
print(f'Difference : {difference:.2f}%')

def analyse_stock(stock):
    reorder_products = []

    for product, details in stock.items():
        quantity = details["qty"]
        reorder_level = details["reorder"]
        price = details["price"]
        cost = details["cost"]

        if quantity <= reorder_level:
            reorder_products.append(product)

        gross_margin = ((price - cost) / price) * 100

        print(f"{product}: Gross margin = {gross_margin:.2f}%")

    return reorder_products


result = analyse_stock(stock)

# Calculate stock values
total_costs = sum(
    details["qty"] * details["cost"]
    for details in stock.values()
)

potential_revenue = sum(
    details["qty"] * details["price"]
    for details in stock.values()
)

gross_profit = potential_revenue - total_costs

gross_margin = (gross_profit / potential_revenue) * 100

# Stall Performance Report
print("=" * 40)
print("       STALL PERFORMANCE REPORT")
print("=" * 40)

print(f"Total Sales:          KES {total:,.2f}")
print(f"Average Daily Sales:  KES {avg:,.2f}")
print(f"Stock Cost Value:     KES {total_costs:,.2f}")
print(f"Potential Revenue:    KES {potential_revenue:,.2f}")
print(f"Potential Gross Profit: KES {gross_profit:,.2f}")
print(f"Gross Margin:         {gross_margin:.2f}%")

print("-" * 40)
print(f"Week 1 Sales:         KES {week1:,.2f}")
print(f"Week 2 Sales:         KES {week2:,.2f}")
print(f"Weekly Growth:        {growth_percentage:.2f}%")
print("=" * 40)
```
